# Import our libraries

```python
import os
import json
import pandas as pd
```

# Load our data

```python
# List files in the root directory
root_files = os.listdir('../')

print(root_files)
```

```python
reviews_path = os.path.join('../', '01-review-collecting')
os.listdir(reviews_path)
```

```python
file_path = os.path.join(reviews_path, 'backpack_reviews.json')

with open(file_path, 'r') as json_file:
    data = json.load(json_file)
```

```python
data
```

```python
all_reviews = []
for page in data.values():
    all_reviews.extend(page)
```

```python
all_reviews
```

```python
df = pd.DataFrame(all_reviews)
df.head()
```

```python
df_reviews = df[['uuid', 'title', 'body']]
df_reviews.head()
```
